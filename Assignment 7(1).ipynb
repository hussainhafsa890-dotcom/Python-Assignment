```python

class Employee:

    def __init__(self, name, number, address, salary, post,
                 year_of_joining, education, year_of_experience):

        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.year_of_experience = year_of_experience

    # Display employee details
    def describe_employee(self):
        print("\n----- Employee Details -----")
        print("Name:", self.name)
        print("Number:", self.number)
        print("Address:", self.address)
        print("Salary:", self.salary)
        print("Post:", self.post)
        print("Year of Joining:", self.year_of_joining)
        print("Education:", self.education)
        print("Year of Experience:", self.year_of_experience)

    # Update experience
    def update_year_of_experience(self, years):
        self.year_of_experience = years
        print("Updated Experience:", self.year_of_experience)

    # Increase salary
    def increase_salary(self, amount):
        self.salary += amount
        print("Updated Salary:", self.salary)


# Creating employee101
employee101 = Employee(
    "Ali", 101, "Karachi", 50000,
    "Manager", 2020, "BS Computer Science", 5
)

# Printing four attributes individually
print("Name:", employee101.name)
print("Number:", employee101.number)
print("Address:", employee101.address)
print("Salary:", employee101.salary)

# Calling all methods
employee101.describe_employee()
employee101.increase_salary(10000)
employee101.update_year_of_experience(6)


# Creating 10 different employee instances

employee1 = Employee("Ahmed", 102, "Karachi", 45000, "Accountant", 2021, "B.Com", 4)

employee2 = Employee("Sara", 103, "Lahore", 60000, "Developer", 2019, "BSCS", 6)

employee3 = Employee("Usman", 104, "Islamabad", 55000, "Designer", 2022, "BS Design", 3)

employee4 = Employee("Ayesha", 105, "Karachi", 70000, "HR Manager", 2018, "MBA", 7)

employee5 = Employee("Hamza", 106, "Multan", 40000, "Assistant", 2023, "BBA", 2)

employee6 = Employee("Zara", 107, "Lahore", 65000, "Teacher", 2020, "M.Ed", 5)

employee7 = Employee("Bilal", 108, "Karachi", 80000, "Team Lead", 2017, "BSCS", 8)

employee8 = Employee("Hira", 109, "Peshawar", 50000, "Developer", 2021, "BSIT", 4)

employee9 = Employee("Saad", 110, "Quetta", 35000, "Clerk", 2024, "B.Com", 1)

employee10 = Employee("Maham", 111, "Karachi", 75000, "Manager", 2019, "MBA", 6)


# Calling describe_employee() for all 10 instances

employee1.describe_employee()
employee2.describe_employee()
employee3.describe_employee()
employee4.describe_employee()
employee5.describe_employee()
employee6.describe_employee()
employee7.describe_employee()
employee8.describe_employee()
employee9.describe_employee()
employee10.describe_employee()


# Updating salaries

employee1.increase_salary(5000)
employee2.increase_salary(10000)
employee3.increase_salary(7000)
employee4.increase_salary(12000)
employee5.increase_salary(3000)


# Updating years of experience

employee6.update_year_of_experience(6)
employee7.update_year_of_experience(9)
employee8.update_year_of_experience(5)
employee9.update_year_of_experience(2)
employee10.update_year_of_experience(7)
```
